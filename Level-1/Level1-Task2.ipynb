```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset .csv")
```

```python
city_count = df['City'].value_counts()
city_count.head()
```

```python
top_city_by_restaurants = city_count.idxmax()
top_city_by_restaurants
```

```python
city_avg_rating = df.groupby('City')['Aggregate rating'].mean()
city_avg_rating.head()
```

```python
top_city_by_rating = city_avg_rating.idxmax()
top_city_by_rating
```

```python
result_city_table = pd.DataFrame({
    "City": city_count.index,
    "Number of Restaurants": city_count.values
}).head(10)

result_city_table
```

```python
plt.figure()
result_city_table.set_index("City")["Number of Restaurants"].plot(kind='bar')
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.title("Top Cities by Number of Restaurants")
plt.show()
```

```python
top_rated_cities = city_avg_rating.sort_values(ascending=False).head(10)

plt.figure()
top_rated_cities.plot(kind='bar')
plt.xlabel("City")
plt.ylabel("Average Rating")
plt.title("Top Cities by Average Restaurant Rating")
plt.show()
```
