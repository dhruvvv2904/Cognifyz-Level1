```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset .csv")
```

```python
df['Price range'].head()
```

```python
price_count = df['Price range'].value_counts().sort_index()
price_count
```

```python
total_restaurants = len(df)

price_percentage = (price_count / total_restaurants) * 100
price_percentage
```

```python
price_table = pd.DataFrame({
    "Price Range": price_count.index,
    "Number of Restaurants": price_count.values,
    "Percentage (%)": price_percentage.round(2).values
})

price_table
```

```python
plt.figure()
plt.bar(price_table["Price Range"], price_table["Number of Restaurants"])
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Distribution of Restaurant Price Ranges")
plt.show()
```
