```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset .csv")
```

```python
df['Has Online delivery'].value_counts()
```

```python
delivery_count = df['Has Online delivery'].value_counts()
total_restaurants = len(df)

delivery_percentage = (delivery_count / total_restaurants) * 100
delivery_percentage
```

```python
delivery_percentage_table = pd.DataFrame({
    "Online Delivery": delivery_count.index,
    "Number of Restaurants": delivery_count.values,
    "Percentage (%)": delivery_percentage.round(2).values
})

delivery_percentage_table
```

```python
avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
avg_rating_delivery
```

```python
rating_comparison_table = avg_rating_delivery.reset_index()
rating_comparison_table.columns = ["Online Delivery", "Average Rating"]

rating_comparison_table
```

```python
plt.figure()
plt.bar(
    rating_comparison_table["Online Delivery"],
    rating_comparison_table["Average Rating"]
)
plt.xlabel("Online Delivery Availability")
plt.ylabel("Average Rating")
plt.title("Average Rating: Online Delivery vs No Online Delivery")
plt.show()
```
